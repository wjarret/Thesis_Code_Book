```python
#Sliding window
def window(data,window_size,step):
    d = {}
    for key, df in data.items():
        n = len(df)
        for i in range(0,n-window_size+1, step):
            window = df.iloc[i:window_size+i].reset_index(drop = True)
            identity = f'{key}_{i}:{window_size+i}'
            d[identity]=window
    return(d)

test_set_dictionary = {k: pd.DataFrame(v) for k, v in test_set_dictionary.items()}
Test_set = window(test_set_dictionary, 24, 1)
len(Test_set)

Test_set_df = pd.concat(Test_set.values(), keys=Test_set.keys())
labels_for_columns = ['Pos. 1', 'Pos. 2', 'Pos. 3', 'Pos. 4', 'Pos. 5', 'Pos. 6', 'Pos. 7', 'Pos. 8', 'Pos. 9', 'Pos. 10', 'Pos. 11', 'Pos. 12', 'Pos. 13', 'Pos. 14', 'Pos. 15', 'Pos. 16', 'Pos. 17', 'Pos. 18', 'Pos. 19', 'Pos. 20','Pos. 21','Pos. 22','Pos. 23','Pos. 24']
Test_set_df = Test_set_df.pivot_table(index=Test_set_df.index.get_level_values(0), columns=Test_set_df.index.get_level_values(1))
Test_set_df.columns = labels_for_columns

Test_set_df['SM_Site'] = 0
Test_set_df.loc['U11.1_80:104', 'SM_Site'] = 1
Test_set_df.loc['U11.2_80:104', 'SM_Site'] = 1
Test_set_df.loc['U11.3_78:102', 'SM_Site'] = 1

Test_set_df[Test_set_df['SM_Site'] == 1]
```
